```python
import pandas as pd
import re
from io import StringIO

# 1. Load the data by reading lines directly to ensure each line is a single entry
with open('/content/drive/MyDrive/连环画/Ref002/Author_Process.csv', 'r', encoding='utf-8') as f:
    lines = f.readlines()

df = pd.DataFrame(lines, columns=['Author'])
# Strip any leading/trailing whitespace, including newline characters
df['Author'] = df['Author'].str.strip()

def clean_authors(text):
    if pd.isna(text) or not isinstance(text, str):
        return text

    # regex to find "Role：Names"
    # Supports both standard colon (:) and Chinese colon (：)
    pairs = re.findall(r'([^,：:]+)[：:]([^,]+)', text)

    # Dictionary to group names by role
    role_dict = {}
    for role, name in pairs:
        role = role.strip()
        name = name.strip()
        if role not in role_dict:
            role_dict[role] = []
        # Add name if not already present for that role
        if name not in role_dict[role]:
            role_dict[role].append(name)

    # Reconstruct the string: "Role:NameName,Role2:NameName"
    output_parts = []
    for role, names in role_dict.items():
        # Join names directly as per your example (关庆留张树文)
        combined_names = "".join(names)
        output_parts.append(f"{role}：{combined_names}")

    return " ".join(output_parts)

# 2. Apply the cleaning function
df['Author'] = df['Author'].apply(clean_authors)

# 3. View results
print(df)

# To save back to a CSV:
df.to_csv('cleaned_authors.csv', index=False)
```
