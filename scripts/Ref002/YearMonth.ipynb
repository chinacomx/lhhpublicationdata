```python
import pandas as pd

# Load your data
df = pd.read_csv('/content/drive/MyDrive/连环画/Ref002/YearMonth.csv')

# Use regex to extract the digits
# (\d+) matches one or more digits
# (?=年) is a lookahead that ensures the digits are followed by '年'
df['Year'] = df['出版年/月'].str.extract(r'(\d+)(?=年)')
df['Month'] = df['出版年/月'].str.extract(r'(\d+)(?=月)')

# Save the result
df.to_csv('YearMonth_Columns.csv', index=False)
```
