```python
# --- Part 2: spaCy Transformer Setup ---
import spacy
# from spacy.util import set_gpu_id # set_gpu_id is deprecated
import warnings

# 3. Install spaCy dependencies for Transformers and GPU (CUDA)
print("\n➡️ Installing spaCy, Transformers, and CUDA dependencies...")
# Install spacy with necessary extras for transformer models and CUDA support
# These have been moved to a separate cell for better organization and to handle
# potential dependency conflicts more easily.

# 4. Download and Load the zh_core_web_trf model
print("➡️ Downloading and loading zh_core_web_trf (Chinese BERT)...")
# Note: This model is large and the download might take a few minutes.
# This has been moved to a separate cell for better organization.

try:
    # 5. Enable GPU for spaCy
    # Use spacy.require_gpu() in newer versions
    spacy.require_gpu(0) # By default, this enables GPU 0.
    print("✅ GPU ID set to 0. spaCy will attempt to use CUDA.")

    # 6. Load the Transformer Model
    nlp = spacy.load("zh_core_web_trf")
    print("✅ spaCy Transformer model loaded: zh_core_web_trf")
    print("This model uses Chinese BERT for high-accuracy NER.")

except Exception as e:
    warnings.warn(f"❌ Could not fully load or initialize GPU. Processing may use CPU. Error: {e}")
    # Fallback to CPU load if GPU setup fails
    nlp = spacy.load("zh_core_web_trf")
    print("⚠️ Loaded model, but forced to CPU mode. Check setup if speed is slow.")

# Re-run the rest of the analysis code (Step 3) after this block.
```

```python
# --- Part 3: Extraction Function and Application ---

def extract_names_and_non_names(text):
    """
    Analyzes a text string to extract personal names (PERSON) using spaCy
    and identifies the remaining components as non-name data.
    """
    # Handle missing or empty data
    if pd.isna(text) or str(text).strip() == "":
        return [], ""

    text = str(text).strip()
    doc = nlp(text)
    names = []

    # 1. Extract PERSON entities
    for ent in doc.ents:
        # We only care about explicitly identified PERSON entities
        if ent.label_ == "PERSON":
            names.append(ent.text.strip())

    # De-duplicate names
    unique_names = list(set(names))

    # 2. Extract non-name data (residual text)
    non_name_text = text

    # Remove all extracted names from the original text
    for name in unique_names:
        # Use regex substitution to remove the name, escaping special characters
        # The 're.sub' replaces the name with a space to help separate remaining tokens
        non_name_text = re.sub(re.escape(name), ' ', non_name_text)

    # Clean up common separators and excessive whitespace in the residual text
    # This leaves only the meaningful non-name words (e.g., '等', '作者', or missed names)
    non_name_text = re.sub(r'[,，、;；\s]+', ' ', non_name_text).strip()
    non_name_text = re.sub(r'\s+', ' ', non_name_text).strip() # Reduce multiple spaces

    return unique_names, non_name_text

# Apply the function to the '编绘' column
print("\n⏳ Starting spaCy analysis and extraction...")

# Create new columns by applying the function
# Note: For 35000 entries and a TRF model, this will take some time even on GPU.
df[['Extracted_Names', 'Non_Name_Data']] = df['编绘'].apply(
    lambda x: pd.Series(extract_names_and_non_names(x))
)

print("✅ Analysis complete.")

# --- Part 4: Review Results and Save to CSV ---

print("\n--- Saving Results to CSV ---")
output_file = 'DatabaseRef004_Processed_NER_Results.csv'
# Select only the relevant columns and save them to a new CSV file
df[['编绘', 'Extracted_Names', 'Non_Name_Data']].to_csv(output_file, index=False, encoding='utf-8')

print(f"✅ Results saved to {output_file}")
# Note: This file is now in your Colab environment and can be manually moved to Drive.

print("\n--- Sample Results (First 10 Rows) ---")
print(df[['编绘', 'Extracted_Names', 'Non_Name_Data']].head(10).to_markdown(index=False))

print("\n--- Summary of Extracted Non-Name Data (Top 10) ---")
# Get and show the top 10 most common non-name phrases
non_name_summary = df['Non_Name_Data'].loc[df['Non_Name_Data'] != ''].value_counts() # Removed .head(100)
print(non_name_summary.head(10).to_markdown(numalign="left", stralign="left")) # Still show top 10 in output for preview
```

## Apply the splitting function

### Subtask:
Apply the splitting function to the '编绘' column to create a new structure (e.g., a list of lists or a new DataFrame) where each original entry is represented as a list of segments.


```python
def split_by_separators(text):
    """
    Splits a text string by common separators like commas, semicolons, and spaces.
    """
    if pd.isna(text) or str(text).strip() == "":
        return []

    # Replace Chinese comma with standard comma before splitting
    text = str(text).strip().replace('、', ',').replace('，', ',')

    # Use regex to split by one or more common separators
    segments = re.split(r'[,;；\s]+', text)

    # Filter out any empty strings resulting from the split
    cleaned_segments = [segment.strip() for segment in segments if segment.strip()]

    return cleaned_segments

# Reapply the splitting function to the '编绘' column with the modified function
df['Segments'] = df['编绘'].apply(split_by_separators)

# Display the head of the DataFrame showing the original '编绘' and the new 'Segments' columns to verify the splitting
print("--- DataFrame head with '编绘' and 'Segments' columns after further correction ---")
print(df[['编绘', 'Segments']].head().to_markdown(index=False))
```

## Define a function to separate name and non-name data per segment

### Subtask:
Create a function that takes a single segment (e.g., "赵伯衡 著") and attempts to identify the personal name and the non-name marker. This might involve using spaCy or other pattern matching techniques.


**Reasoning**:
Define the function to extract names and non-name data from a single segment using spaCy.



```python
def parse_segment(segment):
    """
    Analyzes a single segment string to extract personal names (PERSON) using spaCy
    and identifies the remaining components as non-name data.
    """
    # Handle missing or empty data
    if not isinstance(segment, str) or segment.strip() == "":
        return [], ""

    segment = segment.strip()
    doc = nlp(segment)
    names = []

    # 1. Extract PERSON entities
    for ent in doc.ents:
        # We only care about explicitly identified PERSON entities
        if ent.label_ == "PERSON":
            names.append(ent.text.strip())

    # De-duplicate names
    unique_names = list(set(names))

    # 2. Extract non-name data (residual text)
    non_name_text = segment

    # Remove all extracted names from the original text
    for name in unique_names:
        # Use regex substitution to remove the name, escaping special characters
        # The 're.sub' replaces the name with a space to help separate remaining tokens
        non_name_text = re.sub(re.escape(name), ' ', non_name_text)

    # Clean up common separators and excessive whitespace in the residual text
    non_name_text = re.sub(r'[,，、;；\s]+', ' ', non_name_text).strip()
    non_name_text = re.sub(r'\s+', ' ', non_name_text).strip() # Reduce multiple spaces

    return unique_names, non_name_text

# The function is now defined and ready to be applied to the 'Segments' column.
print("✅ Function 'parse_segment' defined.")
```

## Apply the separation function to each segment

### Subtask:
Iterate through the segments generated in step 2 and apply the separation function from step 3 to each one. Store the results, perhaps in a new structure that links the original entry to the extracted names and markers.


**Reasoning**:
Iterate through each row and each segment within the 'Segments' column, apply the `parse_segment` function, and store the extracted names and non-name data in a structured format.



```python
processed_segments = []

for index, row in df.iterrows():
    row_names = []
    row_non_names = []
    segments = row['Segments']

    # Ensure segments is a list, even if empty
    if not isinstance(segments, list):
        segments = []

    for segment in segments:
        names, non_name_data = parse_segment(segment)
        row_names.extend(names)
        # Keep non-name data as a list of strings, one per segment
        if non_name_data: # Only append if not empty
             row_non_names.append(non_name_data)

    processed_segments.append({'Refined_Names': list(set(row_names)), 'Refined_Non_Name_Data_List': row_non_names})

# Convert the list of dictionaries to a DataFrame
processed_df = pd.DataFrame(processed_segments)

# Join the new DataFrame back to the original DataFrame
df = pd.concat([df, processed_df], axis=1)

# Display the relevant columns to verify the results
print("\n--- DataFrame head with new refined columns ---")
print(df[['编绘', 'Segments', 'Extracted_Names', 'Non_Name_Data', 'Refined_Names', 'Refined_Non_Name_Data_List']].head().to_markdown(index=False))
```

## Consolidate results

### Subtask:
Consolidate the separated names and markers from the 'Refined_Names' and 'Refined_Non_Name_Data_List' columns into a more unified structure, potentially updating the DataFrame with final columns for extracted names and non-name data based on this refined process.


**Reasoning**:
Consolidate the refined names and non-name data into final columns by joining the lists into strings for easier use and display the results.



```python
# 1. Examine the structure and content of the 'Refined_Names' and 'Refined_Non_Name_Data_List' columns.
# (This was done in the previous step and confirmed via the head display)

# 2. Determine the desired final format for the extracted names and non-name data.
# Joining the lists into strings separated by a comma seems appropriate for a final representation in CSV or display.
# For names, we'll join them with ', '. For non-names, we'll also join them with ', '.

# 3. Create new columns in the DataFrame
# Join the lists in 'Refined_Names' into a single string
df['Final_Names'] = df['Refined_Names'].apply(lambda x: ', '.join(x) if isinstance(x, list) else '')

# Keep 'Refined_Non_Name_Data_List' as a list, potentially renaming it for clarity
df['Final_Non_Names_List'] = df['Refined_Non_Name_Data_List']


# 4. Display the head of the DataFrame with the original '编绘' column and the new final columns
print("\n--- DataFrame head with original and final consolidated columns ---")
# Display the list directly for 'Final_Non_Names_List'
print(df[['编绘', 'Final_Names', 'Final_Non_Names_List']].head(10).to_markdown(index=False))

# Optional: Display some rows where non-names were actually found to check the consolidation
print("\n--- Sample rows with consolidated names and non-name lists ---")
# Display the list directly for 'Final_Non_Names_List'
print(df[df['Final_Non_Names_List'].apply(len) > 0][['编绘', 'Final_Names', 'Final_Non_Names_List']].head(10).to_markdown(index=False))
```

```python
from collections import Counter
import pandas as pd

# Flatten the list of lists in 'Final_Non_Names_List' into a single list of all non-names
all_non_names = [item for sublist in df['Final_Non_Names_List'].dropna() for item in sublist if item]

# Count the frequency of each unique non-name phrase
non_name_counts = Counter(all_non_names)

# Convert the Counter object to a pandas Series and sort by frequency
non_name_frequency_series = pd.Series(non_name_counts).sort_values(ascending=False)

# Save the frequency series to a CSV file
output_csv_path_all = 'all_non_name_phrases_frequency.csv'
non_name_frequency_series.to_csv(output_csv_path_all, header=['Frequency'], index_label='Non_Name_Phrase')

print(f"✅ All non-name phrases and their frequencies saved to {output_csv_path_all}")
```
